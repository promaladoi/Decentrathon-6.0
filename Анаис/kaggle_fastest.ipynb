```python
import base64, io, zipfile, pathlib, os, sys, subprocess
root=pathlib.Path('/kaggle/working/anais_fastest')
assert not root.exists(), 'Restart session to preserve existing measured evidence'
root.mkdir()
with zipfile.ZipFile(io.BytesIO(base64.b64decode(''))) as z:
    z.extractall(root)
os.chdir(root/'anais-evaluation')
subprocess.run([sys.executable,'task/prepare_data.py'],check=True)
subprocess.run([sys.executable,'-m','pytest','solution','-q'],check=True)
subprocess.run([sys.executable,'harness/run_task.py','--repeats','2'],check=True)
checks=subprocess.run([sys.executable,'harness/check_rules.py'])
with zipfile.ZipFile('/kaggle/working/anais-fastest-evidence.zip','w',zipfile.ZIP_DEFLATED) as z:
    for p in pathlib.Path('runs').rglob('*'):
        if p.is_file() and not {'triton_cache','inductor_cache','adapter'}.intersection(p.parts):
            z.write(p,p.as_posix())
assert checks.returncode == 0, 'Some checks failed; read runs/checks.json'
```
