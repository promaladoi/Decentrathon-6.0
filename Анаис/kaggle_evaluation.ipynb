```python
import subprocess, sys
p = subprocess.run(['nvidia-smi', '--query-gpu=name', '--format=csv,noheader'], capture_output=True, text=True, check=True)
print(p.stdout)
assert 'T4' in p.stdout, 'Select GPU T4 x2 before running'
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'soup-cli[train]==0.77.0', 'transformers==5.16.1', 'peft==0.20.0', 'trl==0.29.1', 'bitsandbytes==0.50.2', 'accelerate==1.14.0', 'pytest'], check=True)
```

```python
import base64, io, zipfile, pathlib, os
payload = ''
root = pathlib.Path('/kaggle/working/anais_bundle')
if root.exists():
    raise RuntimeError('Project folder already exists. Restart the session for a fresh run; existing results have been preserved.')
root.mkdir()
with zipfile.ZipFile(io.BytesIO(base64.b64decode(payload))) as archive:
    for entry in archive.infolist():
        entry.filename = entry.filename.replace(chr(92), '/')
        destination = (root / entry.filename).resolve()
        if not destination.is_relative_to(root.resolve()):
            raise ValueError('Unsafe archive path')
        archive.extract(entry, root)
repo = root / 'anais-evaluation'
assert (repo / 'harness/run_task.py').is_file()
os.chdir(repo)
print('Project:', repo)
```

```python
import unittest
suite = unittest.defaultTestLoader.discover('anais', pattern='test_report.py')
assert unittest.TextTestRunner(verbosity=2).run(suite).wasSuccessful(), 'Report tests failed'
subprocess.run([sys.executable, 'task/prepare_data.py'], check=True)
subprocess.run([sys.executable, '-m', 'pytest', 'solution', '-q'], check=True)
```

```python
subprocess.run([sys.executable, 'harness/run_task.py', '--repeats', '2'], check=True)
```

```python
checks = subprocess.run([sys.executable, 'harness/check_rules.py'])
print('Official checks exit code:', checks.returncode)
subprocess.run([sys.executable, 'harness/score.py', 'runs/result.json'], check=True)
subprocess.run([sys.executable, 'anais/benchmark_report.py', '--result', 'runs/result.json', '--out', 'anais/reports/latest'], check=True)
from IPython.display import display, Markdown
display(Markdown(pathlib.Path('anais/reports/latest/REPORT.md').read_text(encoding='utf-8')))
```

```python
import zipfile
output = pathlib.Path('/kaggle/working/anais-results.zip')
with zipfile.ZipFile(output, 'w', zipfile.ZIP_DEFLATED) as archive:
    for folder in ('runs', 'anais/reports'):
        for path in pathlib.Path(folder).rglob('*'):
            if path.is_file() and not any(p in ('triton_cache', 'inductor_cache') for p in path.parts) and path.suffix != '.safetensors':
                archive.write(path, path.as_posix())
from IPython.display import FileLink
display(FileLink(str(output)))
```
